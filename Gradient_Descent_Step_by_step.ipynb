```python
from  sklearn.datasets import make_regression
import numpy as np
```

```python
X, y = make_regression(n_samples = 4, n_features = 1, n_informative = 1, n_targets = 1, noise = 80, random_state = 13)
```

```python
#note- no of data is 4, input have 1 feature means one column, output using this feature create 1 output, single column output, noise is 80
```

```python
#Lets apply ordinary least Square
#import LinearRegression
from sklearn.linear_model import LinearRegression
```

```python
#Train the Model
linear_reg = LinearRegression()
linear_reg.fit(X, y)
```

```python
plt.scatter(X, y)
plt.plot(X, linear_reg.predict(X), color = "red")
```

```python
#Lets apply Gradient Descent assuming slope is constant m = 78.35
#and Let's assume the starting value for intercept b =  0
#reshape-Change 2D-array to 1D-array
y_pred = ((78.35 * X) + 100).reshape(4)
```

```python
plt.scatter(X, y)
plt.plot(X, linear_reg.predict(X), color = 'red', label = 'OLS')
plt.plot(X, y_pred, color = 'green', label = 'b = 0')
plt.legend()
plt.show()
```

```python
#slope is fixed
m = 78.35
b = 100
#Calculate Slope or derivative
loss_slope = -2 * np.sum(y - m*X.ravel() - b)
loss_slope
```

```python
#Lets take learning rate = 0.1
lr = 0.1

step_size = loss_slope * lr
print(step_size)
```

```python
#Calculating the new intercept
b = b - step_size
print(b)
```

```python
y_pred1 = ((78.35 * X) + b).reshape(4)

plt.scatter(X, y)
plt.plot(X, linear_reg.predict(X),color = 'red', label = 'OLS')
plt.plot(X, y_pred1, color = 'green', label = 'b = {}'.format(b))
plt.plot(X, y_pred, color = '#A3E4D7', label = 'b = 0')
plt.legend()
plt.show()
```
